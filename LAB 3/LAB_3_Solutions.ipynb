````python
class py_solution:

    def pow(self,a,b):
        if b==0:
            return 1
        if b<0:
            res=1
            for i in range(-b):
                res=res/a
        else:
            res=1
            for i in range(b):
                res=res*a
        return res



# test_case

print(py_solution().pow(2, -3));        # Output  = 0.125  
print(py_solution().pow(3, 5));         # Output = 243
print(py_solution().pow(100, 0));       # Output = 1
````

Write a Python class that takes operations as mentioned below:

Given 2 arrays p1 and p2, the operation p1&p2 should return `True` if all elements in both the array are same else should return `False`. <br>

Given 2 arrays p1 and p2, the operation p1|p2 should return `True` if any of the elements in both the array are same else should return `False`. <br>

**Sample**
```
p1=MyClass([1,2,3])
p2=MyClass([1,2,3])
print(p1&p2) # output: True
print(p1|p2) # output: True

p1=MyClass([1,2,5])
p2=MyClass([1,2,3])
print(p1&p2) # output: False
print(p1|p2) # output: True

p1=MyClass([1,0,3])

print(p1&p2) # output: False
print(p1|p2) # output: False

```


<font color="red">Note: You should not use any inbuilt functions like `all` and `any`</font>


````python
class MyClass:
    def __init__(self,nums):
        self.nums=nums
    def __and__(self,other):
        for i,j in zip(self.nums,other.nums):
            if i!=j:return False
        return True
    def __or__(self,other):
        for i,j in zip(self.nums,other.nums):
            if i==j:return True
        return False

p1=MyClass([1,2,3])
p2=MyClass([1,2,3])
print(p1&p2) # output: True
print(p1|p2) # output: True

p1=MyClass([1,2,5])
p2=MyClass([1,2,3])
print(p1&p2) # output: False
print(p1|p2) # output: True

p1=MyClass([1,0,3])
p2=MyClass([2,4,7])
print(p1&p2) # output: False
print(p1|p2) # output: False

````

**Write a Python class to check the validity of a string of parentheses,`'(', ')', '{', '}', '[' and '].` These brackets must be closed in the correct order, for example `"()"` and `"()[]{}"` are valid but `"[)"`, `"({[)]"` and `"{{{"` are invalid**

````python

class py_class:
  def is_valid_parenthese(self, str1):
       # write your code here.
    brackets = []
    for bracket in str1:
      if bracket in ["(", "{", "["]:
        brackets.append(bracket)
      else:
        if len(brackets) == 0:
          return False
        if bracket == ")" and brackets[-1] != "(":
          return False
        if bracket == "}" and brackets[-1] != "{":
          return False
        if bracket == "]" and brackets[-1] != "[":
          return False
        brackets.pop()
              
    return not brackets


# test_cases

print(py_class().is_valid_parenthese("(){}[]"))
print(py_class().is_valid_parenthese("()[{)}"))
print(py_class().is_valid_parenthese("()"))
````

## Question 2:


**Write a Python class to get all possible unique subsets from a set of distinct integers.**

Example:

Input:

`test_list = [4, 5, 6]`

Output:

`[[], [6], [5], [5, 6], [4], [4, 6], [4, 5], [4, 5, 6]]`

````python
class py_class:
    def sub_sets(self, sset):
      
      # write your code here
      subsets = [[]] # initialize with empty subset
      for num in sset:
          new_subsets = []
          for subset in subsets:
              new_subset = subset + [num]
              if new_subset not in subsets:
                  new_subsets.append(new_subset)
          subsets += new_subsets
      return subsets


# test_case

print(py_class().sub_sets([4,5,6]))
````

## Question 3:

**Write a Python class to find the three elements that sum to zero from a set of n real numbers.**


Example:

Input: `test_list = [-25, -10, -7, -3, 2, 4, 8, 10]`

Output: `[[-10, 2, 8], [-7, -3, 10]]`


````python
class py_class:
    def sum_to_zero(self, nums):
        #write your code here
        n = len(nums)
        if n < 3:
            return []

        triplets = []
        for i in range(n-2):
            for j in range(i+1, n-1):
                for k in range(j+1, n):
                    if nums[i] + nums[j] + nums[k] == 0:
                        triplets.append([nums[i], nums[j], nums[k]])

        return triplets
        







# test_case

print(py_class().sum_to_zero([-25, -10, -7, -3, 2, 4, 8, 10]))
````

## Question 7:

**Write the definition of a Point class. Objects from this class should have a**

*a method `show` to display the coordinates of the point*

*a method `move` to change these coordinates.*

*a method `dist` that computes the distance between 2 points.*

1.png



````python
import math

class Point(object):

    #:param x: the value on the X-axis
    #:param y: the value on the Y-axis
      
    def __init__(self, x, y):
        
        self.x = x
        self.y = y

    # write your code here
    def show(self):
      print('x coordinate : ', self.x)
      print('y coordinate :', self.y)

    def move(self, dx, dy):
        self.x += dx
        self.y += dy

    def dist(self, other):
        dx = self.x - other.x
        dy = self.y - other.y
        return math.sqrt(dx**2 + dy**2)
# test_case

p1 = Point(2, 3)
p2 = Point(3, 3)

p1.show()
p1.move(10, -10)
p1.show()

p2.show()

print(p1.dist(p2))
````
