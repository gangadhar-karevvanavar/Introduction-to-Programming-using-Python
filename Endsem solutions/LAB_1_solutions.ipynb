## Question 5: Volume of n-dimensional Sphere

Volume of `n`-dimensional sphere of radius `R` can computed using following pattern,
4.png

Volume of `n = 1` dimensional sphere of radius `R` (line)  =  $2R$

Volume of `n = 2` dimensional sphere of radius `R` (cicle) = $ \pi R^2$

Volume of `n  = 3` dimensional sphere of radius `R` (ball)  = (volume of 1 dimensional sphere) $ \times \; (2\pi /  n) \times R^2$

Volume of `n  = 4` dimensional sphere of radius `R`  = (volume of 2 dimensional sphere) $ \times \; (2\pi /  n) \times R^2$

Volume of `n  = 5` dimensional sphere of radius `R`  = (volume of 3 dimensional sphere) $ \times \; (2\pi /  n) \times R^2$

and so on ...

<font color='red'>**Write a function that takes input of radius `R` and dimension `n` and return sphere volume**

Hint: Do not hard code (compute mannually) as in image, your code must give answer for any `n` and `R`. Use control statements (loops, if ele), list to store all results and print for required input. Set variable `pi = 3.1415` and use in your code.


```python
def volume_of_n_sphere(R, n):
    # write your code here
    if n == 1:
      volume = (2*R)
    elif n == 2:
      volume = 3.1415*R*R
    else:
      volume = ((2*3.1415*R*R)/n)*volume_of_n_sphere(R, n-2)
    return volume

V = volume_of_n_sphere(1, 7)      # test_case
print(V)
```

Example:

`d1 = {'w': 50, 'x': 100, 'y': 'Green', 'z': 400}`

## Question 5: Difference of consecutives


<font color='red'>**Write function that takes a list as input and returns a list containing absolute differences of consecutive elements in input list**</font>

Example:

Input: `[3, 1, 21, 7, 2, 5, -1, 4, 11, 3]`

Output: `[2, 20, 14, 5, 3, 6, 5, 7, 8 ]`

-----------------------------------------------

Input: `[5, 2, 1, 11, 3, 6, 4, 0]`

Output: `[3, 1, 10, 8, 3, 2, 4 ]`

--------------------------

Input: `[1, 5, 4, 3, 7, 17, 2]`

Output: `[4, 1, 1, 4, 10, 15]`


```python
def consecutive_diff(lst):
    # write your code here
    outlist = []
    for i in range(1, len(lst)):
      diff = lst[i] - lst[i - 1]
      if diff >= 0:
        outlist.append(diff)
      else:
        outlist.append((-1)*diff)
    return outlist
    

test_list = [11, 2, -1, 8, 13, 5, 5, 7, 4, 9]    # test_case
diff_list = consecutive_diff(test_list)

print(diff_list)
```

## Question 5: Duplicate Characters




<font color='red'>**Write a function to print all duplicate characters in a given string**</font>

Examples:

Input: `'hello'`

Output: `l`

-------------------------------------

Input:  `'abracadabra'`

Output = `a, b, r`

--------------------------------------------

Input: `indian`

Output: `i, n`



```python
def duplicate_letters(string):
    # write  your code here.
    letter_count = {}
    for letter in string:
        if letter in letter_count:
            letter_count[letter] = letter_count[letter] + 1
        else:
            letter_count[letter] = 1
    
    duplicate_letters = []
    for letter, count in letter_count.items():
        if count > 1:
            duplicate_letters.append(letter)
    
    if duplicate_letters:
        print("Duplicate letters in the string are: ", end='')
        print(*duplicate_letters)





input_string = 'statistics'


duplicate_letters(input_string)

```

### Question 7

Write a python program to find maximum 2 numbers in a list

example:

Input: `[23, 11, 55, 34, 78, 02, 42, 89, 32]`

output: `89, 78`


```python

def find_two_max_numbers(input_list):
    max1 = max(input_list[0], input_list[1])
    max2 = min(input_list[0], input_list[1])
    for i in range(2, len(input_list)):
        if input_list[i] > max1:
            max2 = max1
            max1 = input_list[i]
        elif input_list[i] > max2:
            max2 = input_list[i]
    return (max1, max2)

input_list = [23, 11, 55, 34, 78, 2, 42, 89, 32]
result = find_two_max_numbers(input_list)
print("Result: The two maximum numbers are", result[0], "and", result[1])

```

```python


def unique(mylist):

    out_list = []

    for j in mylist:

        if j not in out_list:
            out_list.append(j)

    return out_list



input_list = [1,2,3,3,3,3,4,5]

output_list = unique(input_list)

print(output_list)
```
