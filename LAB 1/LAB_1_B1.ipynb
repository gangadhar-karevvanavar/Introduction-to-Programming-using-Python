## Question 3: Letters Counting

<font color='red'>**Write a function to print number of count of each distinct letters in a word (string).**

Example:

Input: `string  = 'abracadabra'`

Output: 

`a 5`

`b 2`

`r 2`

`c 1`

`d 1`


Hint: create a dictionary count = {'a': 1, 'b': 1, 'o': 2, 'n': 1, 's': 2, 'r': 2, 'u': 2, 't': 1} usign for loop and if else conditions and print keys and values

Input: `string = 'india'`

Output: 

`i 2`

`n 1`

`d 1`

`a 1`




```python
def count_letters(string):
    # write your code here
    dic = {}
    for c in string:
      if c in dic:
        dic[c] += 1
      else:
        dic[c] = 1
    for key in dic:
      print(key, dic[key])

test_string = 'Technology'  # test_case

count_letters(test_string)
```

## Question 4: Join Tuples 


<font color='red'>**Write a python program to joint multiple tuples in a list if they have similar initial element**

Example:

Input  : `test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)] `

Output :` [(5, 6, 7, 8), (6, 10), (7, 13)] `

Input: `test_list = [(1, 2), (2, 5), (2, 1, 4), (7, 6)]`

Output: `[(1, 2), (2, 5, 1, 4), (7,6)]`





```python
test_list = [(20, 10), (10, 40), (20, 10, 5), (50, 80), (10, 90)]       # test_case

# write your code here

combined_dict = {}
for tup in test_list:
    if tup[0] in combined_dict:
        combined_dict[tup[0]] = combined_dict[tup[0]] + tup[1:]
    else:
        combined_dict[tup[0]] = tup
print(list(combined_dict.values()))
```

## Question 5: Volume of n-dimensional Sphere

Volume of `n`-dimensional sphere of radius `R` can computed using following pattern,
4.png

Volume of `n = 1` dimensional sphere of radius `R` (line)  =  $2R$

Volume of `n = 2` dimensional sphere of radius `R` (cicle) = $ \pi R^2$

Volume of `n  = 3` dimensional sphere of radius `R` (ball)  = (volume of 1 dimensional sphere) $ \times \; (2\pi /  n) \times R^2$

Volume of `n  = 4` dimensional sphere of radius `R`  = (volume of 2 dimensional sphere) $ \times \; (2\pi /  n) \times R^2$

Volume of `n  = 5` dimensional sphere of radius `R`  = (volume of 3 dimensional sphere) $ \times \; (2\pi /  n) \times R^2$

and so on ...

<font color='red'>**Write a function that takes input of radius `R` and dimension `n` and return sphere volume**

Hint: Do not hard code (compute mannually) as in image, your code must give answer for any `n` and `R`. Use control statements (loops, if ele), list to store all results and print for required input. Set variable `pi = 3.1415` and use in your code.


```python
def volume_of_n_sphere(R, n):
    # write your code here
    if n == 1:
      volume = (2*R)
    elif n == 2:
      volume = 3.1415*R*R
    else:
      volume = ((2*3.1415*R*R)/n)*volume_of_n_sphere(R, n-2)
    return volume

V = volume_of_n_sphere(2, 17)      # test_case
print(V)
```
