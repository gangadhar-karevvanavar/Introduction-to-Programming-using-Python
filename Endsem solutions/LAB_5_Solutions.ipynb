## Question 3. Complex numbers
Create a class `ComplexNumber` with attributes `real` and `imaginary`. The class should have overloading for operators `+`,`-`,`*` and comparators
`==` , `>` , `<`. Also overload `__str__` to print in format `<real> + <imaginary>i`

Note: Comapare using magnitude of complex number.


````python
class ComplexNumber():
  # write your code for __init__ and other required methods.
  def __init__(self,real,imaginary):
        self.real = real
        self.imaginary = imaginary
  def __add__(self,other):
      return ComplexNumber(self.real+other.real,self.imaginary+other.imaginary)
  def __sub__(self,other):
      return ComplexNumber(self.real-other.real,self.imaginary-other.imaginary)
  def __mul__(self,other):
      return ComplexNumber(self.real*other.real-self.imaginary*other.imaginary,self.real*other.imaginary+self.imaginary*other.real)
  def __eq__(self,other):
      return self.real == other.real and self.imaginary == other.imaginary
  def __gt__(self,other):
      return self.real > other.real and self.imaginary > other.imaginary
  def __lt__(self,other):
      return self.real < other.real and self.imaginary < other.imaginary
  def __str__(self) -> str:
      return str(self.real)+" + "+str(self.imaginary)+"i"


# Testcases
c1 = ComplexNumber(1,2)
c2 = ComplexNumber(2,3)
print(c1+c2) # 3 + 5i
print(c1-c2) # -1 - 1i
print(c1*c2) # -4 + 7i
print(c1==c2) # False
print(c1>c2) # False
print(c1<c2) # True
````

## Question 4: Union of 2 lists
Create a class `MyList` which should inherit all methods of inbuild class `list` but should have overload on addition (`+`) operator which should return a new list which is union of two lists.


````python
class MyList(list):
  # write your code for add operator
  def __add__(self,other):
    ret = list(self)
    for i in range(len(other)):
      if other[i] not in self:
        ret.append(other[i])
    return ret


# Testcases
l1 = MyList([1,2,3])
l2 = MyList([2,4,5])
print(l1+l2) # [1, 2, 3, 4, 5]
````

````python
class Vector():
  def __init__(self, A:list):
    # write your code here
    self.A = A

  def __mul__(self, other):
    if len(self.A) != len(other.A):
      print("Dimmensions not matched")
      return 
    ans = 0
    for i in range(len(self.A)):
      ans += (self.A[i]*other.A[i])
    return ans

  def __add__(self, other):
    if len(self.A) != len(other.A):
      print("Dimmensions not matched")
      return 
    ans = []
    for i in range(len(self.A)):
      ans.append(self.A[i]+other.A[i])
    return ans
  
  def __sub__(self, other):
    if len(self.A) != len(other.A):
      print("Dimmensions not matched")
      return
    ans = []
    for i in range(len(self.A)):
      ans.append(self.A[i]-other.A[i])
    return ans # or you can create a new instance of Vector and return it.


# Testcases

v1 = Vector([1,2,3])
v2 = Vector([4,5,6])
print(v1*v2) # 32
print(v1+v2) # [5,7,9]
print(v1-v2) # [-3,-3,-3]
````

### Question 3. Co-ordinates
Create a class `Point` which represents a point in 3 dimmensional geometry i.e. the class has attributes `x`,`y` and `z` representing the x,y,z coordinates of the point. Overload the addition(`+`) and subtraction(`-`) operators such that addition of two point classes (with attribute values `x1`, `y1`, `z1` and `x2`, `y2`, `z2`) should return a new `Point` class with atrribute values `x1+x2`,`y1+y2` and `z1+z2`, subtraction of two point classes (with attribute values `x1`, `y1`, `z1` and `x2`, `y2`, `z2`) should return a new `Point` class with atrribute values `x1-x2`,`y1-y2` and `z1-z2`.

````python
class Point:
    def __init__(self, x, y, z):
        self.x = x
        self.y = y
        self.z = z
        
    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y, self.z + other.z)
    
    def __sub__(self, other):
        return Point(self.x - other.x, self.y - other.y, self.z - other.z)
````

### Question 4. Matrix multiplication using OOPS

Create a class called `Matrix` which should initialize with a list of lists (2D list). The class should overload the `*` operator to perform matrix multiplication. If the dimmensions of the two matrices are not compatible for multiplication print `Dimmensions not matched`
The class should also overload the addition and subtraction operators to perform matrix addition and subtraction. If the dimmensions of the two matrices donot match print `Dimmensions not matched`

```python
a = Matrix([[1,2,3],[4,5,6]])
b = Matrix([[1,2],[3,4],[5,6]])
print(a*b) # [[22,28],[49,64]]
c = Matrix([[1,2,3],[4,5,6]])
d = Matrix([[7,8,9],[10,11,12]])
print(c+d) # [[8,10,12],[14,16,18]]
print(c-d) # [[-6,-6,-6],[-6,-6,-6]]

```

````python
class Matrix:
    def __init__(self, matrix):
        self.matrix = matrix
    
    def __mul__(self, other):
        if len(self.matrix[0]) != len(other.matrix):
            print("Dimmensions not matched")
            return
        else:
            result = []
            for i in range(len(self.matrix)):
                row = []
                for j in range(len(other.matrix[0])):
                    sum = 0
                    for k in range(len(other.matrix)):
                        sum += self.matrix[i][k] * other.matrix[k][j]
                    row.append(sum)
                result.append(row)
            return result
    
    def __add__(self, other):
        if len(self.matrix) != len(other.matrix) or len(self.matrix[0]) != len(other.matrix[0]):
            print("Dimmensions not matched")
            return
        else:
            result = []
            for i in range(len(self.matrix)):
                row = []
                for j in range(len(self.matrix[0])):
                    row.append(self.matrix[i][j] + other.matrix[i][j])
                result.append(row)
            return result
    
    def __sub__(self, other):
        if len(self.matrix) != len(other.matrix) or len(self.matrix[0]) != len(other.matrix[0]):
            print("Dimmensions not matched")
            return
        else:
            result = []
            for i in range(len(self.matrix)):
                row = []
                for j in range(len(self.matrix[0])):
                    row.append(self.matrix[i][j] - other.matrix[i][j])
                result.append(row)
            return result

## Test cases
m = Matrix([[1,2,3],[4,5,6]])
n = Matrix([[1,2],[3,4],[5,6]])
print(m*n) # [[22, 28], [49, 64]]
print(m+m) # [[2, 4, 6], [8, 10, 12]]
print(m-m) # [[0, 0, 0], [0, 0, 0]]
print(m+n) # Dimmensions not matched
````

Write a function `my_polygon(n)` that plots a regular polygon with n sides and radius 1. Recall that the radius of a regular polygon is the distance from its centroid to the vertices. Hint: This problem is significantly easier if you think in polar coordinates. Recall that a complete revolution around the unit circle is `2𝜋` radians. 
Note: The first and last point on the polygon should be the point associated with the polar coordinate angles, 0 and 2𝜋, respectively.
Hint: Use np.cos and np.sin on array to get the coordinates of the points 



````python
import matplotlib.pyplot as plt
import numpy as np


def my_polygon(n):
    # write your code here
    
    # calculate the angles for each vertex of the polygon
    angles = np.linspace(0, 2*np.pi, n+1)

    # calculate the x and y coordinates for each vertex of the polygon
    x = np.cos(angles)
    y = np.sin(angles)

    # plot the polygon
    print(x,y)
    plt.plot(x, y)
    plt.axis('equal')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title(f'Regular Polygon with {n} Sides')
    plt.show()
my_polygon(6)
````
