**Shapes inside the plot**

```python
import matplotlib.patches as patches

x = np.arange(1,101)
y1 = np.random.randn(100)*15+range(1,101)
y2 = np.random.randn(100)*15+range(1,101)

fig1 = plt.figure(figsize=(8,8))
ax1 = fig1.add_subplot(111)
plt.plot(x, y1, linestyle ='none', marker='o')

#------------------------------------------------
ax1.add_patch(
patches.Rectangle( (10, 0),         # starting (edge) position of shape
                  50, 50,           # shepe (rectangle) size width and height
                  alpha=0.3,        # transparency level
                  facecolor="red",  # color
                  edgecolor="black", #edge color
                  linewidth=3,       # line size
                  linestyle='solid')) # line style

ax1.add_patch( patches.Circle((80, 90),         # to plot circle, circle center postion
                              radius = 20,      # radius of circle
                              alpha=0.3, 
                              facecolor="green", 
                              edgecolor="black", 
                              linewidth=1, 
                              linestyle='solid'))

#--------------------------------------------------
plt.show()
```

**Correlation Matrix**

A correlation matrix is simply a table which displays the correlation coefficients for different variables. 

The matrix depicts the correlation between all the possible pairs of values in a table

```python
x_value = ["t1","t2","t3","t4","t5"]
y_value = np.random.random((100,5))

#--------------------------------------------
heat_mat_df = pd.DataFrame(y_value, 
                columns=x_value)

correlation_matrix = heat_mat_df.corr()         #generate correlation matrix

sns.heatmap(correlation_matrix, cmap='PuOr')    # customize cmap color
#--------------------------------------------------
```

**Random Normal: Gaussian distribution**

```python
mu, sigma = 0, 0.1
gaussian_data = np.random.normal(mu, sigma, 1000)   #Generate radomd data points with Gaussian distribution
                                                    #define mean(mu) and standard deviation(sigma)

#-----------------------------------------------------------------
count, bins, ignored = plt.hist(gaussian_data, 30, density=True)
plt.plot(bins, 
         1/(sigma * np.sqrt(2 * np.pi)) *np.exp( - (bins - mu)**2 / (2 * sigma**2)),    #plot gaussian distribution
         linewidth=2, color='r')

#------------------------------------------------------------------
plt.show()
```

## Donut Plot

```python

plt.rcParams["figure.figsize"] = (20,7)


#---------------------------------------------------
circle_partition_data = [15, 9, 5, 35]
data_names = ['data_1', 'data_2', 'data_3', 'data_4']

# To plot the donut plot, first plot the pie chart and plot another white circle inside the pie chart

plt.pie(circle_partition_data,          
        labels=data_names,  
        labeldistance=1.15,
        wedgeprops = { 'linewidth' : 3, 'edgecolor' : 'white' },
        );                      

mask_circle=plt.Circle( (0,0), 0.7, color='white')  # plot another white circle inside the pie chart
p=plt.gcf()
p.gca().add_artist(mask_circle) # mask the pie chart with white circle

#--------------------------------------------------------

plt.show()
```

## Radar / Spider Plot

```python
categories = ['data_1', 'data_2', 'data_3', 'data_4', 'data_5']
values = [38.0, 29, 8, 7, 28, 38.0]     # radar plot values in each direction 
angles = [0.0, 1.25, 2.51, 3.76, 5.02, 0.0] # at which angles varibles should plot

ax = plt.subplot(111, polar=True)
#-------------------------------------------------------
plt.xticks(angles[:-1], categories, color='grey', size=8)       # plot values in angles

ax.set_rlabel_position(0)
plt.yticks([10,20,30], ["10","20","30"], color="grey", size=7)
plt.ylim(0,40)

ax.plot(angles, values, 
        linewidth=1, 
        linestyle='solid', 
        label = 'Event_1')  #Plot all values in respective angles

ax.fill(angles, values, 'b', alpha=0.1)
#-----------------------------------------------------------------
plt.legend(loc='upper right', bbox_to_anchor=(0.4, 0.1))


plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import networkx as nx   # import network library to plot network graphs
```

**Basic Network**

# Animation

```python
import matplotlib.pyplot as plt 
import matplotlib.animation as animation 
import numpy as np 
from IPython.display import Image

fig = plt.figure(figsize=(7,5)) 
ax = plt.axes(xlim=(-50, 50), ylim=(-50, 50)) 
line_draw, = ax.plot([], [], lw=2) 

#--------------------------------------
def initiate():                         #animation intiation function, defining line draws sets
	line_draw.set_data([], []) 
	return line_draw, 

x_data, y_data = [], []                 # the data to be animate, x_data and y_data
x = np.arange(-100,101)
y1 = np.random.randn(200)*10+range(-100,100)


#---------------------------------------------------
def animate_plot(i):                # animation plot function, create the data required to plot animate 
    t = i                           # by defining the time t, and appending x_data and y_data
    x_value = x[t]
    y_value = y1[t]
    x_data.append(x_value)
    y_data.append(y_value)
    line_draw.set_data(x_data, y_data)
    return line_draw, 



plt.title('Animation : Company Share Price')    # animation plot title
plt.axis('off') 

anim = animation.FuncAnimation(fig, animate_plot,       # use function to be animate
                               init_func=initiate,      # initiation function
							    frames=200,             # total number of frames to be displayed
                                interval=20,            # interval between each frame
                                blit=True) 

#---------------------------------------------------------------
anim.save('Animation.gif', writer = "pillow", fps=20 )      # save plottted animation as gif file


Image(open('Animation.gif','rb').read())            # Read the gif file using Image, and show the result.
```

```python
import matplotlib.pyplot as plt 
import matplotlib.animation as animation 
import numpy as np 
from IPython.display import Image


fig = plt.figure(figsize=(7,5)) 
ax = plt.axes(xlim=(-50, 50), ylim=(-50, 50)) 
line_draw, = ax.plot([], [], lw=2) 

#---------------------------------------
def initiate(): 
	line_draw.set_data([], []) 
	return line_draw, 

x_data, y_data = [], [] 

#------------------------------------------
def animate_plot(i):
    t = 0.1*i
    x_value = t
    y_value = t*np.cos(t)           # changing the x_data and y_data, here function of animation is exponetial cos function
    x_data.append(x_value)
    y_data.append(y_value)
    line_draw.set_data(x_data, y_data)
    return line_draw, 
 
plt.title('Animation using Matplotlib: Cos function') 
plt.axis('off') 

anim = animation.FuncAnimation(fig, animate_plot, 
                               init_func=initiate, 
							    frames=500, 
                                interval=20, 
                                blit=True) 

#_----------------------------------------------------------

anim.save('Animation.gif', writer = "pillow", fps=20 )


Image(open('Animation.gif','rb').read())
```
