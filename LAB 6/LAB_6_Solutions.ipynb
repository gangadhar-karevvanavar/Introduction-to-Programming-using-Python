## Question 3: Intersection of 2 dictionaries
**Create a class `MyDict` which should inherit all methods of inbuild class `dict` but should have overload on addition (`+`) operator which should return a new dict which is intersection of two dictionaries.**

````python
class MyDict(dict):
    def __add__(self, other):
        intersection = {}
        for key in self.keys():
            if key in other:
                intersection[key] = self[key]
        return MyDict(intersection)

# Testcases
d1 = MyDict({1: "hi"})
d2 = MyDict({1: "hi", 2: "bye"})
print(d1+d2) # output {1: 'hi'}
````

**Write a program to compute the distance from current position to its starting position (origin).**

*Note: write code for general case*

Hint: Update current coordinates after each move, e.g, for (UP, 5) - updated coordinate is (0, 5), update till last move and compute distance from last coordinates to origin using distance equation. 

3.png



````python
import math

moves = [('UP', 5), ('LEFT', 3), ('DOWN', 3), ('RIGHT', 4), ('DOWN', 1), ('RIGHT', 2), ('UP', 3), ('LEFT', 5)] # testcase

# Initialize current coordinates to (0,0)
current_x = 0
current_y = 0

# Iterate through the list of moves
for move in moves:
    direction = move[0]
    steps = move[1]
    
    # Update the current coordinates based on the direction and number of steps
    if direction == 'UP':
        current_y += steps
    elif direction == 'DOWN':
        current_y -= steps
    elif direction == 'LEFT':
        current_x -= steps
    elif direction == 'RIGHT':
        current_x += steps

# Calculate the distance between the current coordinates and the starting point (0,0)
distance = math.sqrt(current_x**2 + current_y**2)
print(distance)
````

## Question 3: Finding solutions using Cramer's rule

**Complete the function ```cramer()``` below for a $3 \times 3$ matrix such that you return the output as a list [$x1$, $x2$, $x3$]. Make sure to use numpy functions to calculate determinants.** 

Here is a visual representation of the *Cramer rule*: \\

cramer.png

**Hint** : ```temp[:,i] = b``` replaces the $i^{th}$ column of temp matrix with $b$.

````python
import numpy as np

def cramer(coeff, B):
    det_A = np.linalg.det(coeff)
    det_Ax = np.linalg.det(np.column_stack((B, coeff[:,1], coeff[:,2])))
    det_Ay = np.linalg.det(np.column_stack((coeff[:,0], B, coeff[:,2])))
    det_Az = np.linalg.det(np.column_stack((coeff[:,0], coeff[:,1], B)))
    x = det_Ax/det_A
    y = det_Ay/det_A
    z = det_Az/det_A
    return [x, y, z]


# Define coefficients matrix
coeff = np.array([[2, 1, -1],
                  [3, 2, 1],
                  [1, -1, 2]])    # matrix of coefficients

# Define the constants matrix
B = np.array([8, 14, 6])    # d1, d2, d3


cramer(coeff, B) # output x, y, z = [4.399, 0, 0.799] 
````

## Question 4: Longest Common Prefix

**Write code in the ```lcp()``` function to find the longest common prefix**.

**Example**: \\

```
Input: ["flower", "flow", "flight"] \\
Output: "fl"
```

````python
def lcp(words):
    if not words:
        return ""
    prefix = words[0]
    for i in range(1, len(words)):
        while words[i].find(prefix) != 0:
            prefix = prefix[:len(prefix)-1]
            if not prefix:
                return ""
    return prefix




words = ["abc121", "abcdefg", "abcdxyz", "abcdABCD"]
print(lcp(words)) # output abc
````

## Question 3: Grouping Anagrams


**Write a python program for given an array of strings ```strs```, that group the anagrams together.** 

Note : You can return the answer in any order.

*An Anagram is a word or phrase formed by rearranging the letters of a different word or phrase, typically using all the original letters exactly once.* \\

**Example** \\
```
Input: strs = ["eat", "tea", "tan", "ate", "nat", "bat"]
Output: [ ["bat"], ["nat", "tan"], ["ate", "eat", "tea"] ]
```

Complete the ```anagrams()``` function to achieve the above.

````python
from collections import defaultdict

def anagrams(strs):
    anagram_groups = defaultdict(list)
    
    for string in strs:
        sorted_string = ''.join(sorted(string))
        
        anagram_groups[sorted_string].append(string)
    
    return list(anagram_groups.values())


strs = ["eat", "tea", "tan", "ate", "nat", "bat"]
anagrams(strs) # Print [["bat"], ["nat", "tan"], ["ate", "eat", "tea"]]
````

## Question 4: Sum Of Digits

**Complete the ```SumOfDigits``` class to take a number as parameter and on calling the ```sod()``` function, it must return the sum of digits of the number.** 

**Example**:

```SumOfDigits(123).sod()``` returns ```6```. (1 + 2 + 3 = 6)

Further, overload the appropriate operator such that: \\
```SumOfDigits(123).sod()``` < ```SumOfDigits(79).sod()``` returns ```True```

````python
class SumOfDigits:
    def __init__(self, num):
        self.num = num

    def sod(self):
        res = sum(int(d) for d in str(self.num))
        print(res)
        return res

    def __lt__(self, other):
        return self.sod() < other.sod()




n1 = SumOfDigits(123)
n2 = SumOfDigits(79)
sod1 = n1.sod() # Prints 6
sod2 = n2.sod() # Prints 16
print(sod1 < sod2) # Prints True
print(sod2 < sod1) # Prints False
````

## Question 1: Print Butterfly Pattern

butterfly (1).png

Complete the **butterfly** function below to print the pattern shown in the figure.

````python
def butterfly():
  n = 5
  for i in range(n):
      for j in range(n):
          if j <= i:
              print("*", end="")
          else:
              print(" ", end="")
      for j in range(n, 2*n):
          if j >= 2*n-i-1:
              print("*", end="")
          else:
              print(" ", end="")
      print()

  for i in range(n-1, -1, -1):
      for j in range(n):
          if j <= i:
              print("*", end="")
          else:
              print(" ", end="")
      for j in range(n, 2*n):
          if j >= 2*n-i-1:
              print("*", end="")
          else:
              print(" ", end="")
      print()
  pass

butterfly()
````

````python
class Calculator:  
	def __init__(self):  
		self.stack = []
		self.token_map = {
            "+": lambda: self.push(self.pop() + self.pop()),
            "-": lambda: self.push(self.pop() - self.pop()),
            "*": lambda: self.push(self.pop() * self.pop()),
            "/": lambda: self.push(self.pop() / self.pop()),
            "print": lambda: self.print()
        }
		
	def push(self, sth):  
		self.stack.append(int(sth))
		
	def pop(self):
		return self.stack.pop()
		
	def print(self):
		print(self.pop())
		
	def run(self, code):
		tokens = code.split(" ")
		c = 0
		while c < len(tokens):
			current = tokens[c]
			if current.isnumeric():
				self.push(current)
				c+=1
			elif current in self.token_map:
				self.token_map[current]()
				c += 1
			else:
				raise SyntaxError(f"{current}: Invalid Syntax.")
	
calc = Calculator()
calc.run("5 5 + print")
			
````

## Question 6: Valid paranthesis

Complete the functions in class ```ValidParanthesis``` as shown below. The class takes a string ```s``` as parameter and on calling the ```is_valid()``` function, returns ```True``` or ```False``` depending on whether the input string contains valid paranthesis or not. 

Visual representation:

valid.jpg


On encountering an open bracket push into stack (just another list whose elements can be removed from the back of the list - Last In First Out). On encountering a closing bracket, remove the last element from the stack and check if it is the match for the opening bracket. If not, return ```False```. If yes, continue input string traversal and return ```True``` if at the end, the stack is empty. If stack is not empty at the end, return ```False```.

````python
class ValidParenthesis:
    def __init__(self, s):
        self.s = s
        self.stack = []

    def is_valid(self):
        for c in self.s:
            # If the character is an opening bracket, push it onto the stack
            if c in ['(', '[', '{']:
                self.stack.append(c)
            # If the character is a closing bracket, check if it matches the top of the stack
            elif c in [')', ']', '}']:
                # If the stack is empty, return False
                if not self.stack:
                    return False
                # Get the top element from the stack
                top = self.stack.pop()
                # If the brackets don't match, return False
                if (c == ')' and top != '(') or \
                   (c == ']' and top != '[') or \
                   (c == '}' and top != '{'):
                    return False
        # If there are any remaining elements in the stack, return False
        if self.stack:
            return False
        # Otherwise, return True
        return True

s1 = "()[]{}"
vp1 = ValidParenthesis(s1)
print(vp1.is_valid())  # Prints True
````

## Question 7: Best time to buy and sell stock

You are given an array ```prices``` where ```prices[i]``` is the price of a given stock on the $i^{th}$ day. \\

**Example** \\
Input: prices = [7, 1, 5, 3, 6, 4] \\
Output: 5 \\
Explanation: Buy on day 2 (price = 1) and sell on day 5 (price = 6), profit = 6 - 1 = 5.
Note that buying on day 2 and selling on day 1 is not allowed because you must buy before you sell.
You want to maximize your profit by choosing a single day to buy one stock and choosing a different day in the future to sell that stock.

Return the maximum profit you can achieve from this transaction. If you cannot achieve any profit, return 0.

Complete the ```max_profit()``` function to achieve the above.

````python
def max_profit(prices):
    if not prices:
        return 0
    min_price = prices[0]
    max_profit = 0
    for price in prices:
        if price < min_price:
            min_price = price
        elif price - min_price > max_profit:
            max_profit = price - min_price
    return max_profit

prices = [7, 1, 5, 3, 6, 4]
max_profit(prices) # Prints 5
````

## Question 8: Merge Intervals

Complete the ```mergeIntervals()``` function that takes a list of intervals (represented as tuples) and merges overlapping intervals.

Merge Intervals: Write a function 

**Example**:

Input: [(1, 3), (2, 6), (8, 10), (15, 18)] \\
Output: [(1, 6), (8, 10), (15, 18)]

````python
def mergeIntervals(input_list):
      merged = []
      for interval in input_list:
          if not merged or merged[-1][1] < interval[0]:
              merged.append(interval)
          else:
              merged[-1] = (merged[-1][0], max(merged[-1][1], interval[1]))
      return merged

arr = [(1, 3), (2, 6), (8, 10), (15, 18)]
print(mergeIntervals(arr)) # Prints [(1, 6), (8, 10), (15, 18)]
````
