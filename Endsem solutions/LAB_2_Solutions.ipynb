Algorithm Flowchart

Bubble Sort.png


```python
#write your code here
n = 5
arr = []
for i in range(n):
  arr.append(int(input()))
for i in range(n):
  for j in range(0, n-i-1):
    if arr[j] < arr[j+1]:
      arr[j], arr[j+1] = arr[j+1], arr[j]

print(arr)


```

## Question 4: Least Common Multiple

**Write a python program using a algorithm to find `Least Common Multiple (LCM)` of 2 integers.**

Algorithm Flowchart

2.png

```python
# write your code here.
def LCM(x,y):
  z = max(x,y)
  lcm = 0
  while(True):
    if z%x==0 and z%y==0:
      lcm = z
      break;
    z+=1
  return lcm

print(LCM(10,13))



```
